```python
import ray
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
import os

# 1. 環境設定
nb_path = '/content/drive/MyDrive/colab_packages'
if not os.path.exists(nb_path): os.makedirs(nb_path)

if ray.is_initialized():
    ray.shutdown()
ray.init(num_cpus=4)

# ------------------------------------------
# 2. 定義模型 (與之前一致)
# ------------------------------------------
class SecrecyPolicyNet(nn.Module):
    def __init__(self):
        super(SecrecyPolicyNet, self).__init__()
        self.fc = nn.Sequential(
            nn.Linear(6, 64), nn.ReLU(),
            nn.Linear(64, 32), nn.ReLU(),
            nn.Linear(32, 2)
        )
    def forward(self, x): return self.fc(x)

# ------------------------------------------
# 3. ROLL Worker (修正：加入 compute_metrics)
# ------------------------------------------
@ray.remote
class ROLL_6G_Worker:
    def __init__(self, worker_id, num_eves=3):
        self.worker_id = worker_id
        self.num_eves = num_eves
        self.sigma2 = 1e-9 # 雜訊功率

    def compute_metrics(self, r_target, mode='ISAC_UAV', n_uavs=1):
        """論文核心：計算特定場景下的中斷率與能效"""
        # 模擬信道增益 (Channel Gain)
        h_base = np.random.rayleigh(1.2)
        g_eves = np.random.rayleigh(0.6, self.num_eves)

        if mode == 'RIS':
            gain = h_base * 2.5
            inf_eve = np.max(g_eves) * 0.8
            power = 15.0 # BS + RIS 反射功耗
        else: # ISAC_UAV 系列
            gain = h_base * (4.0 + n_uavs * 2.0)
            inf_eve = np.max(g_eves) * (0.2 / n_uavs)
            power = 10.0 + (50.0 * n_uavs) # BS + UAV 飛行與傳輸功耗

        # 計算秘密速率 (Secrecy Rate)
        r_sec = max(0, np.log2(1 + gain/self.sigma2) - np.log2(1 + inf_eve/self.sigma2))

        # 指標 1: 中斷機率 (Outage)
        p_out = 1.0 if r_sec < r_target else 0.01 * np.random.rand() # 加入隨機成功率擾動

        # 指標 2: 能效 (Energy Efficiency)
        ee = (r_sec / power) if p_out < 1.0 else 0.0

        return p_out, ee

# ------------------------------------------
# 4. 執行 ROLL 研究分析函數
# ------------------------------------------
def run_roll_full_study_v2():
    r_targets = [2, 4, 6, 8, 10, 12]
    modes = [('RIS', 0), ('ISAC_UAV', 1), ('ISAC_UAV', 2), ('ISAC_UAV', 3)]
    labels = ['RIS (Passive)', '1-UAV (Relay)', '2-UAV (Relay+Jam)', '3-UAV (Swarm)']

    out_results = {l: [] for l in labels}
    ee_results = {l: [] for l in labels}

    # 部署 3 個遠端 Worker 模擬不同戰場
    workers = {
        "High": ROLL_6G_Worker.remote(worker_id=2, num_eves=3)
    }

    print("📡 ROLL Learner: 開始從邊緣 Worker 收集大規模模擬數據...")

    for rt in r_targets:
        job_ids = []
        for (m, n), l in zip(modes, labels):
            # 這裡調用的是已經定義好的 compute_metrics.remote
            job_ids.append((l, workers["High"].compute_metrics.remote(rt, mode=m, n_uavs=n)))

        for label, j_id in job_ids:
            p_out, ee = ray.get(j_id)
            out_results[label].append(p_out)
            ee_results[label].append(ee)

    # --- 繪圖分析 ---
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
    colors = ['#2ca02c', '#1f77b4', '#ff7f0e', '#d62728']
    markers = ['x', 'o', '^', 's']

    for i, l in enumerate(labels):
        processed_out = [max(x, 1e-4) for x in out_results[l]]
        ax1.semilogy(r_targets, processed_out, color=colors[i], marker=markers[i], label=l, lw=2)
        ax2.plot(r_targets, ee_results[l], color=colors[i], marker=markers[i], label=l, lw=2)

    ax1.set_title('Reliability: Outage Probability', fontsize=14)
    ax1.set_xlabel('Target Secrecy Rate $R_{target}$'); ax1.set_ylabel('$P_{out}$'); ax1.legend(); ax1.grid(True, which='both', ls='--')

    ax2.set_title('Sustainability: Energy Efficiency', fontsize=14)
    ax2.set_xlabel('Target Secrecy Rate $R_{target}$'); ax2.set_ylabel('EE (bits/Joule)'); ax2.legend(); ax2.grid(True, ls='--')

    plt.tight_layout()
    plt.savefig(os.path.join(nb_path, 'ROLL_6G_Full_Comparison_v2.png'), dpi=300)
    plt.show()

# 執行
run_roll_full_study_v2()
```

```python
import ray
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import collections, random

# 1. 初始化 Ray
if ray.is_initialized(): ray.shutdown()
ray.init()

# ------------------------------------------
# 2. ROLL Context Buffer (工業級經驗回放)
# ------------------------------------------
class ROLL_Context_Buffer:
    def __init__(self, capacity=1000):
        self.buffer = collections.deque(maxlen=capacity)

    def push(self, decision, r_sec, p_an, reward):
        self.buffer.append((decision, r_sec, p_an, reward))

    def sample(self, batch_size):
        batch = random.sample(self.buffer, min(len(self.buffer), batch_size))
        decisions, r_secs, an_powers, rewards = zip(*batch)
        return torch.stack(decisions), np.array(r_secs), np.array(an_powers), np.array(rewards)

# ------------------------------------------
# 3. BS Cloud Brain (雲端大腦) & UAV Edge Worker (邊緣執行器)
# ------------------------------------------
class BS_Cloud_Brain(nn.Module):
    def __init__(self, input_dim=12):
        super(BS_Cloud_Brain, self).__init__()
        # 將輸入 (Batch, 12) 視為 (Batch, 1, 12) 的一維信號
        self.conv = nn.Sequential(
            nn.Conv1d(1, 16, kernel_size=3, padding=1), # 提取鄰近特徵關係
            nn.BatchNorm1d(16),
            nn.ReLU(),
            nn.Conv1d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(4) # 縮減維度
        )

        self.fc = nn.Sequential(
           nn.Linear(32 * 4, 64),
           nn.ReLU(),
           nn.Linear(64, 11), # 2 (pos) + 8 (BF: 4 real + 4 imag) + 1 (AN power)
           nn.Tanh()
        )

    def forward(self, x):
        # 增加通道維度: (Batch, 12) -> (Batch, 1, 12)
        if x.dim() == 2:
            x = x.unsqueeze(1)
        x = self.conv(x)
        x = self.fc(x.view(x.size(0), -1))
        return x
@ray.remote(num_cpus=0.1)
class UAV_Edge_Worker:
    def __init__(self, worker_id):
        self.brain_copy = BS_Cloud_Brain()
        self.sigma2 = 1e-9

    def sync_weights(self, new_weights):
        self.brain_copy.load_state_dict(new_weights)

    def collect_and_process(self, uav_pos, target, threat, bs_pos):
        h_sample = np.random.randn(2).astype(np.float32)
        context = torch.FloatTensor(np.hstack([uav_pos, target, threat, bs_pos, h_sample])).unsqueeze(0)

        decision = self.brain_copy(context)
        act = decision.detach().numpy().flatten()

    # 修正切片索引
        move_vec = act[0:2]          # 2
        w_real = act[2:6]            # 4
        w_imag = act[6:10]           # 4
        p_an = np.abs(act[10])       # 1，總計 11

        w_vec = w_real + 1j * w_imag
        w_vec /= (np.linalg.norm(w_vec) + 1e-9)

        dist_uav_tar = np.linalg.norm(uav_pos - target)
        dist_uav_eve = np.linalg.norm(uav_pos - threat)
        h_ue = (np.random.randn(4) + 1j * np.random.randn(4)) / (np.sqrt(2) * (dist_uav_tar + 1.0)**2)
        g_eve = (np.random.randn(4) + 1j * np.random.randn(4)) / (np.sqrt(2) * (dist_uav_eve + 1.0)**2)

        snr_ue = np.abs(np.dot(h_ue, w_vec))**2 / self.sigma2
        snr_eve = np.abs(np.dot(g_eve, w_vec))**2 * 5.0/ (self.sigma2 + p_an)
        r_sec = max(1e-4, np.log2(1 + snr_ue) - np.log2(1 + snr_eve))

        reward = np.log2(1+r_sec) - 0.1 * p_an
        return decision.squeeze(0), r_sec, p_an, reward

# ------------------------------------------
# 4. Asynchronous Learner Logic (CNN 強化避障版)
# ------------------------------------------
dynamic_threshold = 12

def optimize_step(brain, optimizer, buffer, device, batch_size=32, alpha=0.01):
    global dynamic_threshold
    if len(buffer.buffer) < batch_size:
        return None, dynamic_threshold, 0.0, 0.0, 0.0, 0.0

    contexts, r_secs, ans, _ = buffer.sample(batch_size)
    contexts = contexts.to(device)
    r_tensor = torch.FloatTensor(r_secs).to(device)

    # --- EMA 動態門檻 ---
    target_threshold = max(np.percentile(r_secs, 50), 12.0)
    dynamic_threshold = (1 - alpha) * dynamic_threshold + alpha * target_threshold

    brain.train()
    # CNN 輸出不再有 attn_weights
    decs = brain(contexts)

    # EE 計算
    avg_ee = np.mean(r_secs / (1.2 + ans + 1e-9))

# 1. 大幅提升 Secrecy Rate 的獎勵權重 (從 2.5 提升至 5.0)
    # 這是為了讓 CNN 明白：一切移動都是為了這個值
    rate_loss = -torch.mean(r_tensor) * 5.0

    # 2. 強化「中斷懲罰」 (Outage Penalty)
    # 當 R_sec 低於門檻時，給予極大的懲罰值
    outage_penalty = torch.mean(torch.where(
        r_tensor < dynamic_threshold,
        torch.tensor(20.0).to(device), # 提高懲罰，強制 CNN 避開低效區
        torch.tensor(0.0).to(device)
    ))

    # 3. 幾何避障 (防止撞到 Eve)
    dist_to_eve = contexts[:, -1]
    geo_penalty = torch.mean(torch.clamp(6.0 / (dist_to_eve + 0.1)**2, max=15.0))

    # 4. 停留穩定性懲罰：當模型發現 R_sec 很高時，給予移動量 (decs[:, 0:2]) 額外懲罰
    # 這會讓大腦輸出一個與「目標吸引力」抵銷的向量，達成停留效果
    stability_penalty = 0.5 * torch.mean(decs[:, 0:2]**2)

    loss = rate_loss + stability_penalty + outage_penalty + geo_penalty

    optimizer.zero_grad()
    loss.backward()
    torch.nn.utils.clip_grad_norm_(brain.parameters(), max_norm=0.5)
    optimizer.step()

    # 注意：因為移除 Attention，我們回傳 0.0 或其他監控值（如 Loss 變動）
    return {k: v.cpu() for k, v in brain.state_dict().items()}, dynamic_threshold, avg_ee, -loss.item(), loss.item(), 0.0

# ==========================================
# 5. 非同步訓練循環與分析
# ==========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
cloud_brain = BS_Cloud_Brain().to(device)
optimizer = optim.Adam(cloud_brain.parameters(), lr=0.001)
context_buffer = ROLL_Context_Buffer()
edge_workers = [UAV_Edge_Worker.remote(i) for i in range(10)]

initial_weights = {k: v.cpu() for k, v in cloud_brain.state_dict().items()}
ray.get([w.sync_weights.remote(initial_weights) for w in edge_workers])

target_pos = np.array([12.0, 2.0]); threat_pos = np.array([6.0, 5.0]); bs_pos = np.array([0.0, 0.0])
pending_tasks = {edge_workers[i].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos): i for i in range(10)}

history = {'step': [], 'reward': [], 'loss': [], 'r_sec': [], 'threshold': [], 'ee': []}

print("🚀 ROLL 非同步分析系統啟動 (含 Loss/Reward 監測)...")
for step in range(150):
    ready_ids, _ = ray.wait(list(pending_tasks.keys()), num_returns=1)
    for ready_id in ready_ids:
        worker_idx = pending_tasks.pop(ready_id)
        decision, r_sec, p_an, local_reward = ray.get(ready_id)
        context_buffer.push(decision, r_sec, p_an, local_reward)

        new_w, thresh, ee_val, glob_reward, glob_loss, _ = optimize_step(cloud_brain, optimizer, context_buffer, device)

        if new_w is not None:
            edge_workers[worker_idx].sync_weights.remote(new_w)
            history['step'].append(step)
            history['reward'].append(glob_reward)
            history['loss'].append(glob_loss)
            history['r_sec'].append(r_sec)
            history['threshold'].append(thresh)
            history['ee'].append(ee_val)

        new_task = edge_workers[worker_idx].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos)
        pending_tasks[new_task] = worker_idx

# ==========================================
# 6. 三圖專業分析視覺化
# ==========================================
fig = plt.figure(figsize=(22, 6))
gs = gridspec.GridSpec(1, 3)

# 圖 A: 動態風險門檻與秘密速率
ax1 = fig.add_subplot(gs[0])
sma_r_sec = np.convolve(history['r_sec'], np.ones(10)/10, mode='valid')
ax1.plot(history['step'][:len(sma_r_sec)], sma_r_sec, 'b-', label='Secrecy Rate (SMA-10)')
ax1.plot(history['step'], history['threshold'], 'r--', alpha=0.6, label='Risk Threshold')
ax1.set_ylabel('Rate (bits/s/Hz)', color='b')
ax1_twin = ax1.twinx()
ax1_twin.plot(history['step'], history['ee'], 'g-', alpha=0.3, label='Energy Efficiency')
ax1_twin.set_ylabel('EE (bits/J)', color='g')
ax1.set_title('(A) Risk-Aware Performance', fontweight='bold')
ax1.legend(loc='upper left'); ax1.grid(True, ls=':')

# 圖 B: Global Reward 與 Cloud Loss (雙軸)
ax2 = fig.add_subplot(gs[1])
ax2.plot(history['reward'], 'tab:orange', label='Global Reward (Cloud)')
ax2.set_xlabel('Optimization Steps')
ax2.set_ylabel('Reward Value', color='tab:orange')
ax2.tick_params(axis='y', labelcolor='tab:orange')
ax2.legend(loc='upper left')

ax2_twin = ax2.twinx()
ax2_twin.plot(history['loss'], 'tab:gray', linestyle=':', label='Brain Loss (Optimizer)')
ax2_twin.set_ylabel('Loss Value', color='tab:gray')
ax2_twin.tick_params(axis='y', labelcolor='tab:gray')
ax2_twin.legend(loc='upper right')

ax2.set_title('(B) Cloud Brain: Reward vs Loss', fontweight='bold')
ax2.grid(True, ls=':')

# --- 圖 C: 軌跡與速率分析 ---
ax3 = fig.add_subplot(gs[2])
cloud_brain.eval()

uav_p = np.array([1.0, 1.0])
traj = [uav_p.copy()]
r_sec_path = []  # 紀錄飛行過程中的速率
step_size = 0.5

# 模擬計算目前點的 Secrecy Rate (對齊 Worker 邏輯)
def compute_instant_r_sec(uav_pos, target, threat, sigma2=1e-9):
    dist_tar = np.linalg.norm(uav_pos - target)
    dist_eve = np.linalg.norm(uav_pos - threat)
    # 使用平方律路徑損耗讓距離更敏感
    h_ue_gain = 1.0 / (dist_tar + 1.0)**2
    g_eve_gain = 1.0 / (dist_eve + 1.0)**2
    # 簡化版速率估計 (假設平均 Beamforming 增益)
    snr_ue = (1.0 * h_ue_gain) / sigma2
    snr_eve = (1.0 * g_eve_gain) / (sigma2 + 0.1) # 假設 0.1 的人工雜訊
    return max(0, np.log2(1 + snr_ue) - np.log2(1 + snr_eve))

for step_idx in range(100):
    # 計算當前速率
    current_r = compute_instant_r_sec(uav_p, target_pos, threat_pos)
    r_sec_path.append(current_r)

    # 確保 State 輸入
    h_sim = np.array([0.05, -0.05], dtype=np.float32)
    st = torch.FloatTensor(np.hstack([uav_p, target_pos, threat_pos, bs_pos, h_sim])).to(device).unsqueeze(0)

    with torch.no_grad():
        out = cloud_brain(st).cpu().numpy().flatten()

    d_tar = np.linalg.norm(uav_p - target_pos)

    # 1. 引力計算
    attraction_mag = 0.2 * (1 - np.exp(-d_tar/5))
    attraction = (target_pos - uav_p) / (d_tar + 1e-9) * attraction_mag

    # 2. 大腦決策
    brain_decision = out[0:2] * 0.15

    # 3. 總合位移與鎖定邏輯
    move_vec = attraction + brain_decision
    if d_tar < 3.0 and np.linalg.norm(brain_decision) > 0.1:
        uav_p += move_vec * 0.3 # 煞車模式
    else:
        uav_p += move_vec * step_size

    traj.append(uav_p.copy())
    if d_tar < 0.3: break

# --- 繪圖邏輯優化 ---
traj = np.array(traj)
r_sec_path = np.array(r_sec_path)

# 使用顏色代表 Secrecy Rate (顏色越深/越紅代表速率越高)
path_scatter = ax3.scatter(traj[:-1, 0], traj[:-1, 1], c=r_sec_path, cmap='viridis',
                           s=20, edgecolors='none', label='UAV Path (Color: R_sec)')
fig.colorbar(path_scatter, ax=ax3, label='Instant Secrecy Rate (bits/s/Hz)')

# 畫出飛行連線
ax3.plot(traj[:, 0], traj[:, 1], 'g-', alpha=0.3)

ax3.scatter(bs_pos[0], bs_pos[1], c='black', marker='^', s=200, label='BS (Brain)')
ax3.scatter(target_pos[0], target_pos[1], c='b', marker='s', s=100, label='Target')
ax3.scatter(threat_pos[0], threat_pos[1], c='r', marker='x', s=100, label='Eve')

ax3.set_title(f'(C) Trajectory: Final R_sec={r_sec_path[-1]:.2f}', fontweight='bold')
ax3.set_xlim(-2, 15); ax3.set_ylim(-5, 12)
ax3.legend(loc='lower left', fontsize='small'); ax3.grid(True, ls=':')

plt.tight_layout()
plt.show()
save_path = os.path.join(nb_path, 'ROLL_Analysis_CloudCNNUAV.png')
plt.savefig(save_path, dpi=300)
```

```python
import ray
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import collections, random
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# 1. 初始化 Ray
if ray.is_initialized(): ray.shutdown()
ray.init()

# ------------------------------------------
# 2. ROLL Context Buffer (保持不變)
# ------------------------------------------
class ROLL_Context_Buffer:
    def __init__(self, capacity=2000):
        self.buffer = collections.deque(maxlen=capacity)
    def push(self, context, r_sec, p_an, reward):
        self.buffer.append((context, r_sec, p_an, reward))
    def sample(self, batch_size):
        batch = random.sample(self.buffer, min(len(self.buffer), batch_size))
        contexts, r_secs, an_powers, rewards = zip(*batch)
        return torch.cat(contexts), np.array(r_secs), np.array(an_powers), np.array(rewards)

# ------------------------------------------
# 3. 6G PLS 環境 (保持物理邏輯)
# ------------------------------------------
class MIMO6GEnv:
    def __init__(self):
        self.sigma2 = 1e-9
    def step(self, action_tensor, uav_pos, target, threat):
        action = action_tensor.detach().cpu().numpy().flatten()
        w = (action[0:4] + 1j * action[4:8])
        w /= (np.linalg.norm(w) + 1e-9)
        p_an = np.abs(action[8])

        d_tar = np.linalg.norm(uav_pos - target)
        d_eve = np.linalg.norm(uav_pos - threat)

        h_ue = (np.random.randn(4) + 1j * np.random.randn(4)) / (np.sqrt(2) * (d_tar + 1.0)**2)
        g_eve = (np.random.randn(4) + 1j * np.random.randn(4)) / (np.sqrt(2) * (d_eve + 1.0)**2)

        snr_ue = np.abs(np.dot(h_ue, w))**2 / self.sigma2
        snr_eve = np.abs(np.dot(g_eve, w))**2 * 5.0/ (self.sigma2 + p_an)
        r_sec = max(1e-4, np.log2(1 + snr_ue) - np.log2(1 + snr_eve))

        reward = r_sec - 0.2 * p_an
        obs = np.hstack([uav_pos, target, threat, [0,0], [0.1,-0.1], d_tar, d_eve]).astype(np.float32)
        return obs, reward, r_sec, p_an

# ------------------------------------------
# 4. Transformer BS Cloud Brain (修正：激活注意力)
# ------------------------------------------
class BS_Cloud_Brain(nn.Module):
    def __init__(self, input_dim=12, embed_dim=64, nhead=4):
        super().__init__()
        # 修正：將每個特徵映射到 embedding 空間
        self.input_projection = nn.Linear(1, embed_dim)

        self.attention = nn.MultiheadAttention(embed_dim, nhead, batch_first=True)
        self.layernorm = nn.LayerNorm(embed_dim)

        self.output_head = nn.Sequential(
            nn.Flatten(), # 將 (Batch, 12, 64) 展平
            nn.Linear(embed_dim * input_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 9),
            nn.Tanh()
        )

    def forward(self, x, return_weights=False):
        # 核心修正：將 (Batch, 12) 轉為 (Batch, 12, 1) -> 序列長度變為 12
        if x.dim() == 2:
            x = x.unsqueeze(-1)

        # 映射到 Embedding 維度 (Batch, 12, 64)
        x_emb = self.input_projection(x)

        # 這裡的 attn_w 才會是 (Batch, 12, 12) 的矩陣，代表特徵間的互相影響
        attn_out, attn_w = self.attention(x_emb, x_emb, x_emb)

        x_norm = self.layernorm(attn_out + x_emb)
        out = self.output_head(x_norm)

        return (out, attn_w) if return_weights else out

# ------------------------------------------
# 5. Worker & Learner (修正 EE 計算與 Attention 紀錄)
# ------------------------------------------
@ray.remote(num_cpus=0.1)
class UAV_Edge_Worker:
    def __init__(self, worker_id):
        self.env = MIMO6GEnv()
        self.brain_copy = BS_Cloud_Brain()
        # 新增：初始化上一步速率為 0
        self.last_r_sec = 0.0

    def sync_weights(self, weights):
        self.brain_copy.load_state_dict(weights)

    def collect_and_process(self, uav_p, tar, thr, bs):
        d_tar = np.linalg.norm(uav_p - tar)
        d_eve = np.linalg.norm(uav_p - thr)

        # --- 核心修改：將歷史速率資訊嵌入 Context ---
        # 原本的 [0.1, -0.1] 是無意義占位符，現在改為 [當前參考速率, 速率變化趨勢]
        # 讓 Attention 能夠對比：位置維度 vs 速率維度
        ctx_data = np.hstack([
            uav_p, tar, thr, bs,
            [self.last_r_sec, 0.0], # 這裡 0.0 可預留給其他二階特徵
            d_tar, d_eve
        ])
        ctx = torch.FloatTensor(ctx_data).unsqueeze(0)

        with torch.no_grad():
            dec = self.brain_copy(ctx)

        # 執行環境步進
        _, reward, r_sec, p_an = self.env.step(dec, uav_p, tar, thr)

        # --- 更新歷史狀態 ---
        self.last_r_sec = r_sec

        return ctx, r_sec, p_an, reward

def optimize_step(brain, optimizer, buffer, device, batch_size=32, alpha=0.05):
    global dynamic_threshold
    if len(buffer.buffer) < batch_size: return None, dynamic_threshold, 0.0, 0.0, 0.0, 0.0

    contexts, r_secs, ans, _ = buffer.sample(batch_size)
    contexts, r_tensor = contexts.to(device), torch.FloatTensor(r_secs).to(device)

    # 強制門檻維持在高標，不讓它隨波逐流
    target_threshold = max(np.percentile(r_secs, 50), 10.0)
    dynamic_threshold = (1 - alpha) * dynamic_threshold + alpha * target_threshold

    brain.train()
    decs, attn_weights = brain(contexts, return_weights=True)

    # --- 關鍵獎勵函數 ---
    # 1. Secrecy Rate 權重加大 (2.5)，讓它成為 Loss 的主導
    rate_loss = -torch.mean(r_tensor) * 2.5

    # 2. Outage Penalty (解決之前的 NameError)
    outage_penalty = torch.mean(torch.where(r_tensor < dynamic_threshold,
                                           torch.tensor(12.0).to(device), torch.tensor(0.0).to(device)))

    # 3. 幾何避障 (防止撞到 Eve)
    dist_to_eve = contexts[:, -1]
    geo_penalty = torch.mean(torch.clamp(6.0 / (dist_to_eve + 0.1)**2, max=15.0))

    # 4. 停留穩定性懲罰：當模型發現 R_sec 很高時，給予移動量 (decs[:, 0:2]) 額外懲罰
    # 這會讓大腦輸出一個與「目標吸引力」抵銷的向量，達成停留效果
    stability_penalty = 0.5 * torch.mean(decs[:, 0:2]**2)

    loss = rate_loss + stability_penalty + outage_penalty + geo_penalty

    optimizer.zero_grad()
    loss.backward()
    torch.nn.utils.clip_grad_norm_(brain.parameters(), max_norm=0.5)
    optimizer.step()

    avg_ee = np.mean(r_secs / (1.2 + ans + 1e-9))
    attn_val = attn_weights.std().item() * 50
    return {k: v.cpu() for k, v in brain.state_dict().items()}, dynamic_threshold, avg_ee, -loss.item(), loss.item(), attn_val
# ==========================================
# 7. 啟動訓練
# ==========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
cloud_brain = BS_Cloud_Brain().to(device)
optimizer = optim.Adam(cloud_brain.parameters(), lr=0.0001) # 低 LR 1e-4
context_buffer = ROLL_Context_Buffer()
edge_workers = [UAV_Edge_Worker.remote(i) for i in range(10)]

# 初始化同步
initial_weights = {k: v.cpu() for k, v in cloud_brain.state_dict().items()}
ray.get([w.sync_weights.remote(initial_weights) for w in edge_workers])

target_pos = np.array([12.0, 2.0]); threat_pos = np.array([6.0, 5.0]); bs_pos = np.array([0.0, 0.0])
pending_tasks = {edge_workers[i].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos): i for i in range(10)}

history = {'step': [], 'reward': [], 'loss': [], 'r_sec': [], 'threshold': [], 'attn': []}

print("🚀 ROLL Transformer 系統啟動...")
for step in range(150):
    ready_ids, _ = ray.wait(list(pending_tasks.keys()), num_returns=1)
    for ready_id in ready_ids:
        worker_idx = pending_tasks.pop(ready_id)
        context, r_sec, p_an, local_reward = ray.get(ready_id)
        context_buffer.push(context, r_sec, p_an, local_reward)

        res = optimize_step(cloud_brain, optimizer, context_buffer, device)
        if res[0] is not None:
            new_w, thresh, _, glob_reward, glob_loss, attn_val = res
            edge_workers[worker_idx].sync_weights.remote(new_w)
            history['step'].append(step)
            history['reward'].append(glob_reward)
            history['loss'].append(glob_loss)
            history['r_sec'].append(r_sec)
            history['threshold'].append(thresh)
            history['attn'].append(attn_val)

        new_task = edge_workers[worker_idx].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos)
        pending_tasks[new_task] = worker_idx
# ==========================================
# 5. 訓練循環與視覺化
# ==========================================
def compute_path_r_sec(uav_pos, target, threat, sigma2=1e-9):
    # 對齊 MIMO6GEnv 的物理邏輯
    d_tar = np.linalg.norm(uav_pos - target)
    d_eve = np.linalg.norm(uav_pos - threat)

    # 使用平方律路徑損耗
    h_gain = 1.0 / (d_tar + 1.0)**2
    g_gain = 1.0 / (d_eve + 1.0)**2

    # 簡化版 SNR 估計 (假設大腦已優化波束)
    snr_ue = (1.0 * h_gain) / sigma2
    snr_eve = (1.0 * g_gain * 5.0) / (sigma2 + 0.05) # 0.05 為假設的 AN 功率
    return max(0, np.log2(1 + snr_ue) - np.log2(1 + snr_eve))


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
cloud_brain = BS_Cloud_Brain().to(device)
optimizer = optim.Adam(cloud_brain.parameters(), lr=0.001) # 降低 LR 至 1e-4 提升 Transformer 穩定性
context_buffer = ROLL_Context_Buffer()
edge_workers = [UAV_Edge_Worker.remote(i) for i in range(10)]

initial_weights = {k: v.cpu() for k, v in cloud_brain.state_dict().items()}
ray.get([w.sync_weights.remote(initial_weights) for w in edge_workers])

target_pos = np.array([12.0, 2.0]); threat_pos = np.array([6.0, 5.0]); bs_pos = np.array([0.0, 0.0])
pending_tasks = {edge_workers[i].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos): i for i in range(10)}

history = {'step': [], 'reward': [], 'loss': [], 'r_sec': [], 'threshold': [], 'ee': [], 'attn': []}

for step in range(150):
    # 在訓練循環末尾或特定步數
    # 在訓練循環中
    if step % 100 == 0 and len(context_buffer.buffer) >= 32: # 確保有足夠樣本再畫圖
      try:
        attn_map = get_attention_map(cloud_brain, context_buffer, device)
        plot_attention_heatmap(attn_map, step)
        save_path = os.path.join(nb_path, 'ROLL_HEATMAP_CloudTransformerUAV.png')
        plt.savefig(save_path, dpi=300)
      except Exception as e:
        print(f"Heatmap skipped at step {step} due to: {e}")
    ready_ids, _ = ray.wait(list(pending_tasks.keys()), num_returns=1)
    for ready_id in ready_ids:
        worker_idx = pending_tasks.pop(ready_id)
        context, r_sec, p_an, local_reward = ray.get(ready_id)
        context_buffer.push(context, r_sec, p_an, local_reward)

        update_res = optimize_step(cloud_brain, optimizer, context_buffer, device)
        if update_res[0] is not None:
            new_w, thresh, ee_val, glob_reward, glob_loss, attn_val = update_res
            edge_workers[worker_idx].sync_weights.remote(new_w)
            history['step'].append(step)
            history['reward'].append(glob_reward)
            history['loss'].append(glob_loss)
            history['r_sec'].append(r_sec)
            history['threshold'].append(thresh)
            history['ee'].append(ee_val)
            history['attn'].append(attn_val)

        new_task = edge_workers[worker_idx].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos)
        pending_tasks[new_task] = worker_idx
# ==========================================
# HEAT MAP
# ==========================================

import seaborn as sns # 建議安裝此庫用於美化熱圖
def get_attention_map(brain, buffer, device):
    brain.eval()
    # 從 buffer 抽樣一組數據來分析
    contexts, _, _, _ = buffer.sample(32)
    contexts = contexts.to(device)

    with torch.no_grad():
        _, attn_weights = brain(contexts, return_weights=True)

    # attn_weights 形狀通常是 (Batch, Seq_len, Seq_len) -> (32, 12, 12)
    # 我們對 Batch 取平均，得到 12x12 的特徵交互矩陣
    avg_attn = torch.mean(attn_weights, dim=0).cpu().numpy()
    return avg_attn
def plot_attention_heatmap(attn_matrix, step):
    feature_labels = [
        'UAV_x', 'UAV_y', 'Tar_x', 'Tar_y',
        'Eve_x', 'Eve_y', 'BS_x', 'BS_y',
        'Last_R_sec', 'Placeholder', 'Dist_Tar', 'Dist_Eve'
    ]

    plt.figure(figsize=(10, 8))
    sns.heatmap(attn_matrix, annot=True, fmt=".2f", cmap='YlGnBu',
                xticklabels=feature_labels, yticklabels=feature_labels)
    plt.title(f"Transformer Feature Attention Heatmap (Step {step})")
    plt.xlabel("Key (Features being attended to)")
    plt.ylabel("Query (Feature making the decision)")
    plt.xticks(rotation=45)
    plt.show()
# ==========================================
# 6. 三圖專業分析視覺化
# ==========================================
fig = plt.figure(figsize=(22, 6))
gs = gridspec.GridSpec(1, 3)

# 圖 A: 動態風險門檻與秘密速率
ax1 = fig.add_subplot(gs[0])
sma_r_sec = np.convolve(history['r_sec'], np.ones(10)/10, mode='valid')
ax1.plot(history['step'][:len(sma_r_sec)], sma_r_sec, 'b-', label='Secrecy Rate (SMA-10)')
ax1.plot(history['step'], history['threshold'], 'r--', alpha=0.6, label='Risk Threshold')
ax1.set_ylabel('Rate (bits/s/Hz)', color='b')
ax1_twin = ax1.twinx()
ax1_twin.plot(history['step'], history['ee'], 'g-', alpha=0.3, label='Energy Efficiency')
ax1_twin.set_ylabel('EE (bits/J)', color='g')
ax1.set_title('(A) Risk-Aware Performance', fontweight='bold')
ax1.legend(loc='upper left'); ax1.grid(True, ls=':')

# 圖 B: Global Reward, Loss 與 Attention 分佈
ax2 = fig.add_subplot(gs[1])
ax2.plot(history['reward'], 'tab:orange', label='Global Reward', lw=2)
ax2.set_ylabel('Reward', color='tab:orange')
ax2.tick_params(axis='y', labelcolor='tab:orange')

ax2_loss = ax2.twinx()
ax2_loss.plot(history['loss'], 'tab:gray', linestyle=':', label='Brain Loss')
ax2_loss.set_ylabel('Loss', color='tab:gray')

# 新增：Attention 監測（使用第三軸或在原圖標註）
ax2_attn = ax2.twinx()
ax2_attn.spines['right'].set_position(('outward', 60))
ax2_attn.plot(history['attn'], 'tab:purple', alpha=0.5, label='Attention Intensity')
ax2_attn.set_ylabel('Attention Weight', color='tab:purple')
ax2_attn.tick_params(axis='y', labelcolor='tab:purple')

ax2.set_title('(B) Transformer Monitor: Reward, Loss & Attention', fontweight='bold')
ax2.grid(True, ls=':')

# --- 圖 C: 軌跡分析 (含即時速率顯示) ---
ax3 = fig.add_subplot(gs[2])
cloud_brain.eval()

uav_p = np.array([1.0, 1.0])
traj = [uav_p.copy()]
path_r_secs = [] # 新增：紀錄路徑速率
last_r_sim = 15.0

for step_idx in range(80):
    d_tar = np.linalg.norm(uav_p - target_pos)
    d_eve = np.linalg.norm(uav_p - threat_pos)

    # 計算當前點的物理速率
    curr_r = compute_path_r_sec(uav_p, target_pos, threat_pos)
    path_r_secs.append(curr_r)

    # 12 維特徵對齊推理
    st_data = np.hstack([uav_p, target_pos, threat_pos, bs_pos, [curr_r, 0.0], d_tar, d_eve])
    st = torch.FloatTensor(st_data).to(device).unsqueeze(0)

    with torch.no_grad():
        out = cloud_brain(st).cpu().numpy().flatten()

    # 軌跡物理邏輯
    dist_norm = np.linalg.norm(uav_p - target_pos)
    attraction_mag = 0.3 * (1 - np.exp(-dist_norm/5))
    attraction = (target_pos - uav_p) / (dist_norm + 1e-9) * attraction_mag
    brain_decision = out[0:2] * 0.4 # 大腦修正量

    move_vec = attraction + brain_decision

    # 步長控制
    uav_p += move_vec * 0.6
    traj.append(uav_p.copy())

    if dist_norm < 0.4: break

# --- 視覺化優化 ---
traj = np.array(traj[:-1]) # 對齊長度
path_r_secs = np.array(path_r_secs)

# 使用顏色代表 Secrecy Rate (由淺入深代表速率升高)
sc = ax3.scatter(traj[:, 0], traj[:, 1], c=path_r_secs, cmap='viridis',
                 s=30, zorder=3, edgecolors='none', label='UAV Path (R_sec)')
plt.colorbar(sc, ax=ax3, label='Instant Secrecy Rate (bits/s/Hz)')

# 畫出連線
ax3.plot(traj[:, 0], traj[:, 1], 'g-', alpha=0.3)

# 標記關鍵位置
ax3.scatter(bs_pos[0], bs_pos[1], c='black', marker='^', s=200, label='BS (Cloud)')
ax3.scatter(target_pos[0], target_pos[1], c='b', marker='s', s=100, label='Target')
ax3.scatter(threat_pos[0], threat_pos[1], c='r', marker='x', s=100, label='Eve')

ax3.set_title(f'(C) Trajectory Control (Final R={path_r_secs[-1]:.1f})', fontweight='bold')
ax3.set_xlim(-1, 15); ax3.set_ylim(-2, 12)
ax3.legend(loc='lower left', fontsize='small'); ax3.grid(True, ls=':')
plt.tight_layout()
plt.show()
save_path = os.path.join(nb_path, 'ROLL_Analysis_CloudTransformerUAV.png')
plt.savefig(save_path, dpi=300)
```

```python
import ray
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import collections, random
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# 1. 初始化 Ray
if ray.is_initialized(): ray.shutdown()
ray.init()

# ------------------------------------------
# 2. ROLL Context Buffer (保持不變)
# ------------------------------------------
class ROLL_Context_Buffer:
    def __init__(self, capacity=2000):
        self.buffer = collections.deque(maxlen=capacity)
    def push(self, context, r_sec, p_an, reward):
        self.buffer.append((context, r_sec, p_an, reward))
    def sample(self, batch_size):
        batch = random.sample(self.buffer, min(len(self.buffer), batch_size))
        contexts, r_secs, an_powers, rewards = zip(*batch)
        return torch.cat(contexts), np.array(r_secs), np.array(an_powers), np.array(rewards)

# ------------------------------------------
# 3. 6G PLS 環境 支援多 Eve
# ------------------------------------------
class MIMO6GEnv:
    def __init__(self):
        self.sigma2 = 1e-9
    def step(self, action_tensor, uav_pos, target, threats):
        # threats 預期為 [[x1, y1], [x2, y2]]
        action = action_tensor.detach().cpu().numpy().flatten()
        w = (action[0:4] + 1j * action[4:8])
        w /= (np.linalg.norm(w) + 1e-9)
        p_an = np.abs(action[8])

        d_tar = np.linalg.norm(uav_pos - target)
        h_ue = (np.random.randn(4) + 1j * np.random.randn(4)) / (np.sqrt(2) * (d_tar + 1.0)**2)
        snr_ue = np.abs(np.dot(h_ue, w))**2 / self.sigma2

# 計算所有 Eve 的 SNR，取最強（最危險）的一個，或計算總和
        snr_eves = []
        for eve_pos in threats:
          d_eve = np.linalg.norm(uav_pos - eve_pos)
          g_eve = (np.random.randn(4) + 1j * np.random.randn(4)) / (np.sqrt(2) * (d_eve + 1.0)**2)
          snr_e = np.abs(np.dot(g_eve, w))**2* 5.0 / (self.sigma2 + p_an)
          snr_eves.append(snr_e)

        max_snr_eve = max(snr_eves) # 考慮威脅最大者
        r_sec = max(1e-4, np.log2(1 + snr_ue) - np.log2(1 + max_snr_eve))

        reward = r_sec - 0.2 * p_an
        return None,reward, r_sec, p_an

# ------------------------------------------
# 4. Transformer BS Cloud Brain (修正：激活注意力)
# ------------------------------------------
class BS_Cloud_Brain(nn.Module):
    def __init__(self, global_dim=9, eve_dim=3, embed_dim=64, nhead=4):
        """
        global_dim: 固定資訊 (UAV, Tar, BS, Last_R, Dist_Tar) 的長度
        eve_dim: 每個 Eve 的特徵長度 (x, y, dist)
        """
        super().__init__()
        self.embed_dim = embed_dim

        # 1. 獨立的特徵投影層
        self.global_proj = nn.Linear(global_dim, embed_dim)
        self.eve_proj = nn.Linear(eve_dim, embed_dim)

        # 2. Transformer 核心
        self.attention = nn.MultiheadAttention(embed_dim, nhead, batch_first=True)
        self.layernorm = nn.LayerNorm(embed_dim)

        # 3. 輸出頭 (不依賴序列長度)
        self.output_head = nn.Sequential(
            nn.Linear(embed_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 11), # 修正為 11 維輸出 (2 move + 8 BF + 1 AN)
            nn.Tanh()
        )

    def forward(self, x_global, x_eves, return_weights=False):
        """
        x_global: (Batch, global_dim)
        x_eves: (Batch, N_eves, eve_dim) -> N_eves 可以是任何數字
        """
        # --- A. 映射到 Embedding 空間 ---
        # (Batch, 1, embed_dim)
        g_emb = self.global_proj(x_global).unsqueeze(1)
        # (Batch, N_eves, embed_dim)
        e_emb = self.eve_proj(x_eves)

        # --- B. 拼接成 Token 序列 ---
        # 序列形狀: (Batch, 1 + N_eves, embed_dim)
        x_seq = torch.cat([g_emb, e_emb], dim=1)

        # --- C. 注意力計算 ---
        attn_out, attn_w = self.attention(x_seq, x_seq, x_seq)
        x_norm = self.layernorm(attn_out + x_seq)

        # --- D. Global Average Pooling (核心修正) ---
        # 壓縮序列維度 (1 + N_eves)，取平均值。這讓輸出維度永遠是 (Batch, embed_dim)
        # 無論有幾個 Eve，最後都只會輸出一個總結向量
        x_pooled = torch.mean(x_norm, dim=1)

        out = self.output_head(x_pooled)

        return (out, attn_w) if return_weights else out

# ------------------------------------------
# 5. Worker & Learner (修正 EE 計算與 Attention 紀錄)
# ------------------------------------------
@ray.remote(num_cpus=0.1)
class UAV_Edge_Worker:
    def __init__(self, worker_id):
        self.env = MIMO6GEnv()
        self.brain_copy = BS_Cloud_Brain()
        # 新增：初始化上一步速率為 0
        self.last_r_sec = 0.0

    def sync_weights(self, weights):
        self.brain_copy.load_state_dict(weights)

# 建議的新特徵排列 (16維):
# [uav_p(2), tar_p(2), bs_p(2), last_r(1), p_an(1),
#  eve1_p(2), eve2_p(2), eve3_p(2), d_tar(1), d_min_eve(1)]

    def collect_and_process(self, uav_p, tar, threats, bs):
        # 1. 強制將 threats 轉為 2D numpy array (N_eves, 2)
        # 這樣可以防止傳入單個 Eve 時變成 scalar 報錯
        threats = np.array(threats)
        if threats.ndim == 1:
            threats = threats.reshape(-1, 2)

        d_tar = np.linalg.norm(uav_p - tar)

        # 2. 構建 Global 特徵 (x_global)
        g_data = np.hstack([uav_p, tar, bs, [self.last_r_sec, 0.0, d_tar]])
        g_tensor = torch.FloatTensor(g_data).unsqueeze(0).to(device)

        # 3. 構建 Eve 特徵 (x_eves) - 確保每個 token 都是 3 維 [x, y, dist]
        e_list = []
        for i in range(len(threats)):
            e_p = threats[i]
            dist_e = np.linalg.norm(uav_p - e_p)
            # 修正 IndexError: 不要用 e_p[0][0]，直接用 e_p[0]
            e_feat = np.array([e_p[0], e_p[1], dist_e])
            e_list.append(e_feat)

        e_tensor = torch.FloatTensor(np.array(e_list)).unsqueeze(0).to(device)

        # 4. 推理
        with torch.no_grad():
            # 這裡呼叫 BS_Cloud_Brain(x_global, x_eves)
            dec = self.brain_copy(g_tensor, e_tensor)

        # 5. 環境互動 (確保環境也支援 threats 矩陣)
        _, reward, r_sec, p_an = self.env.step(dec, uav_p, tar, threats)

        self.last_r_sec = r_sec

        # 回傳給 Buffer
        return (g_data, np.array(e_list)), r_sec, p_an, reward

dynamic_threshold = 10.0  # 給予一個合理的初始值
def optimize_step(brain, optimizer, buffer, device, batch_size=32, alpha=0.05):
    global dynamic_threshold
    # 1. 安全檢查
    if len(buffer.buffer) < batch_size:
        return None, dynamic_threshold, 0.0, 0.0, 0.0, 0.0

    # 1. 從 buffer 取樣
    batch = random.sample(buffer.buffer, batch_size)

    # 2. 第一次解包：對應 Worker 的 return (context, r_sec, p_an, reward)
    # 注意：這裡我們只接收 4 個變數
    contexts, r_secs, ans, rewards = zip(*batch)

    # 3. 第二次解包：拆開 context 裡的 (g_data, e_list)
    # 這是因為 Worker 存入時把 global 和 eves 封裝成了一個 tuple
    g_list, e_list = zip(*contexts)

    # 4. 轉換為 Tensor
    g_tensor = torch.FloatTensor(np.array(g_list)).to(device)    # (Batch, 9)
    e_tensor = torch.FloatTensor(np.array(e_list)).to(device)    # (Batch, N, 3)
    r_tensor = torch.FloatTensor(np.array(r_secs)).to(device)

    # 3. 動態門檻計算 (維持高標)
    target_threshold = max(np.percentile(r_secs, 50), 10.0)
    dynamic_threshold = (1 - alpha) * dynamic_threshold + alpha * target_threshold

    brain.train()
    # 使用雙輸入呼叫新模型
    decs, attn_weights = brain(g_tensor, e_tensor, return_weights=True)

    # --- 關鍵損失函數計算 ---

    # A. Secrecy Rate 獎勵 (加權加大，強化專注度)
    rate_loss = -torch.mean(r_tensor) * 2.5

    # B. Outage Penalty (中斷懲罰)
    outage_penalty = torch.mean(torch.where(
        r_tensor < dynamic_threshold,
        torch.tensor(15.0).to(device), # 提高懲罰力度
        torch.tensor(0.0).to(device)
    ))

    # C. 動態幾何避障 (針對所有 Eve)
    # e_tensor 的最後一維假設是距離 (dist)
    # 形狀為 (Batch, N_eve, 3)，其中 [:, :, 2] 是距離
    dist_to_eves = e_tensor[:, :, 2]
    # 對所有 Eve 的壓迫感求和
    geo_penalty_tensor = torch.clamp(7.0 / (dist_to_eves + 0.1)**2, max=20.0)
    geo_penalty = torch.mean(torch.sum(geo_penalty_tensor, dim=1))

    # D. 停留穩定性懲罰
    # 當 R_sec 很高時，懲罰不必要的移動量 decs[:, 0:2]
    stability_penalty = 0.6 * torch.mean(decs[:, 0:2]**2)

    # 總 Loss
    loss = rate_loss + stability_penalty + outage_penalty + geo_penalty

    # 4. 優化步驟
    optimizer.zero_grad()
    loss.backward()
    torch.nn.utils.clip_grad_norm_(brain.parameters(), max_norm=0.5)
    optimizer.step()

    # 5. 監控指標
    avg_ee = np.mean(np.array(r_secs) / (1.2 + np.array(ans) + 1e-9))
    attn_val = attn_weights.std().item() * 50

    return ({k: v.cpu() for k, v in brain.state_dict().items()},
            dynamic_threshold, avg_ee, -loss.item(), loss.item(), attn_val)
# ==========================================
# 7. 啟動訓練
# ==========================================
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
cloud_brain = BS_Cloud_Brain().to(device)
optimizer = optim.Adam(cloud_brain.parameters(), lr=0.0001) # 低 LR 1e-4
context_buffer = ROLL_Context_Buffer()
edge_workers = [UAV_Edge_Worker.remote(i) for i in range(10)]

# 初始化同步
initial_weights = {k: v.cpu() for k, v in cloud_brain.state_dict().items()}
ray.get([w.sync_weights.remote(initial_weights) for w in edge_workers])

target_pos = np.array([12.0, 2.0]); threat_pos = np.array([6.0, 5.0]); bs_pos = np.array([0.0, 0.0])
pending_tasks = {edge_workers[i].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos): i for i in range(10)}

history = {'step': [], 'reward': [], 'loss': [], 'r_sec': [], 'threshold': [], 'attn': []}

print("🚀 ROLL Transformer 系統啟動...")
for step in range(150):
    ready_ids, _ = ray.wait(list(pending_tasks.keys()), num_returns=1)
    for ready_id in ready_ids:
        worker_idx = pending_tasks.pop(ready_id)
        context, r_sec, p_an, local_reward = ray.get(ready_id)
        context_buffer.push(context, r_sec, p_an, local_reward)

        res = optimize_step(cloud_brain, optimizer, context_buffer, device)
        if res[0] is not None:
            new_w, thresh, _, glob_reward, glob_loss, attn_val = res
            edge_workers[worker_idx].sync_weights.remote(new_w)
            history['step'].append(step)
            history['reward'].append(glob_reward)
            history['loss'].append(glob_loss)
            history['r_sec'].append(r_sec)
            history['threshold'].append(thresh)
            history['attn'].append(attn_val)

        new_task = edge_workers[worker_idx].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos)
        pending_tasks[new_task] = worker_idx
# ==========================================
# 5. 訓練循環與視覺化
# ==========================================
# 新增：隨機生成 Eve 的函數
def generate_random_threats(num_min=2, num_max=3):
    num_eves = np.random.randint(num_min, num_max + 1)
    # 在 4~10 範圍內隨機生成多個 Eve 座標
    return np.random.uniform(4, 10, (num_eves, 2))

# --- 修改訓練啟動部分 ---
target_pos = np.array([12.0, 2.0])
bs_pos = np.array([0.0, 0.0])

# 初始任務啟動 (每個 Worker 面對的 Eve 數量和位置都不同)
pending_tasks = {}
for i in range(10):
    uav_init = np.random.uniform(0, 4, 2)
    current_threats = generate_random_threats(2, 3) # 隨機 2 或 3 個
    task = edge_workers[i].collect_and_process.remote(uav_init, target_pos, current_threats, bs_pos)
    pending_tasks[task] = i

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
cloud_brain = BS_Cloud_Brain().to(device)
optimizer = optim.Adam(cloud_brain.parameters(), lr=0.001) # 降低 LR 至 1e-4 提升 Transformer 穩定性
context_buffer = ROLL_Context_Buffer()
edge_workers = [UAV_Edge_Worker.remote(i) for i in range(10)]

initial_weights = {k: v.cpu() for k, v in cloud_brain.state_dict().items()}
ray.get([w.sync_weights.remote(initial_weights) for w in edge_workers])

target_pos = np.array([12.0, 2.0]); threat_pos = np.array([6.0, 5.0]); bs_pos = np.array([0.0, 0.0])
pending_tasks = {edge_workers[i].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos): i for i in range(10)}

history = {'step': [], 'reward': [], 'loss': [], 'r_sec': [], 'threshold': [], 'ee': [], 'attn': []}

for step in range(250):
    # 在訓練循環末尾或特定步數
    ready_ids, _ = ray.wait(list(pending_tasks.keys()), num_returns=1)
    for ready_id in ready_ids:
        worker_idx = pending_tasks.pop(ready_id)
        context, r_sec, p_an, local_reward = ray.get(ready_id)
        context_buffer.push(context, r_sec, p_an, local_reward)

        update_res = optimize_step(cloud_brain, optimizer, context_buffer, device)
        if update_res[0] is not None:
            new_w, thresh, ee_val, glob_reward, glob_loss, attn_val = update_res
            edge_workers[worker_idx].sync_weights.remote(new_w)
            history['step'].append(step)
            history['reward'].append(glob_reward)
            history['loss'].append(glob_loss)
            history['r_sec'].append(r_sec)
            history['threshold'].append(thresh)
            history['ee'].append(ee_val)
            history['attn'].append(attn_val)

        new_task = edge_workers[worker_idx].collect_and_process.remote(np.random.uniform(0,4,2), target_pos, threat_pos, bs_pos)
        pending_tasks[new_task] = worker_idx

# ==========================================
# 6. 三圖專業分析視覺化
# ==========================================
def get_brain_action(brain, uav_p, target_pos, threat_pos_list, bs_pos, last_r_sim, device):
    """封裝推理邏輯，確保縮進安全並對齊動態 Eve 結構"""
    brain.eval()
    d_tar = np.linalg.norm(uav_p - target_pos)

    # 1. 構建 Global 特徵 (9維)
    # [uav_x, uav_y, tar_x, tar_y, bs_x, bs_y, r_sec, dummy, d_tar]
    g_data = np.hstack([uav_p, target_pos, bs_pos, [last_r_sim, 0.0, d_tar]])
    g_tensor = torch.FloatTensor(g_data).to(device).unsqueeze(0)

    # 2. 構建動態 Eve 特徵 (N_eves, 3維)
    # 每個 Eve 為 [x, y, dist]
    e_list = []
    for e_p in threat_pos_list:
        dist_e = np.linalg.norm(uav_p - e_p)
        e_list.append(np.hstack([e_p, [dist_e]]))
    e_tensor = torch.FloatTensor(np.array(e_list)).to(device).unsqueeze(0)

    # 3. 推理 (在函數內縮進極其安全)
    with torch.no_grad():
        out, attn_w = brain(g_tensor, e_tensor, return_weights=True)
        out = out.cpu().numpy().flatten()

    return out, attn_w

fig = plt.figure(figsize=(22, 6))
gs = gridspec.GridSpec(1, 3)

# 圖 A: 動態風險門檻與秘密速率
ax1 = fig.add_subplot(gs[0])
sma_r_sec = np.convolve(history['r_sec'], np.ones(10)/10, mode='valid')
ax1.plot(history['step'][:len(sma_r_sec)], sma_r_sec, 'b-', label='Secrecy Rate (SMA-10)')
ax1.plot(history['step'], history['threshold'], 'r--', alpha=0.6, label='Risk Threshold')
ax1.set_ylabel('Rate (bits/s/Hz)', color='b')
ax1_twin = ax1.twinx()
ax1_twin.plot(history['step'], history['ee'], 'g-', alpha=0.3, label='Energy Efficiency')
ax1_twin.set_ylabel('EE (bits/J)', color='g')
ax1.set_title('(A) Risk-Aware Performance', fontweight='bold')
ax1.legend(loc='upper left'); ax1.grid(True, ls=':')

# 圖 B: Global Reward, Loss 與 Attention 分佈
ax2 = fig.add_subplot(gs[1])
ax2.plot(history['reward'], 'tab:orange', label='Global Reward', lw=2)
ax2.set_ylabel('Reward', color='tab:orange')
ax2.tick_params(axis='y', labelcolor='tab:orange')

ax2_loss = ax2.twinx()
ax2_loss.plot(history['loss'], 'tab:gray', linestyle=':', label='Brain Loss')
ax2_loss.set_ylabel('Loss', color='tab:gray')

# 新增：Attention 監測（使用第三軸或在原圖標註）
ax2_attn = ax2.twinx()
ax2_attn.spines['right'].set_position(('outward', 60))
ax2_attn.plot(history['attn'], 'tab:purple', alpha=0.5, label='Attention Intensity')
ax2_attn.set_ylabel('Attention Weight', color='tab:purple')
ax2_attn.tick_params(axis='y', labelcolor='tab:purple')

ax2.set_title('(B) Transformer Monitor: Reward, Loss & Attention', fontweight='bold')
ax2.grid(True, ls=':')

# --- 圖 C: 軌跡分析 (動態速率顯示版本) ---
ax3 = fig.add_subplot(gs[2])
cloud_brain.eval()

uav_p = np.array([1.0, 1.0])
traj = [uav_p.copy()]
path_r_secs = [] # 儲存飛行過程中的即時速率
last_r_sim = 18.0

# 確保 threat_pos_list 已定義
if 'threat_pos_list' not in locals():
    num_eves = np.random.randint(2, 4)
    threat_pos_list = [np.random.uniform(3, 11, 2) for _ in range(num_eves)]

for step_idx in range(100):
    # 1. 執行推理
    out, _ = get_brain_action(cloud_brain, uav_p, target_pos, threat_pos_list, bs_pos, last_r_sim, device)

    # 2. 計算目前位置的即時速率 (對齊 MIMO6GEnv 物理邏輯)
    d_tar = np.linalg.norm(uav_p - target_pos)
    dists_eves = [np.linalg.norm(uav_p - e) for e in threat_pos_list]
    min_d_eve = min(dists_eves)

    # 使用平方律路徑損耗模擬 (對應您觀察到的速率斷層現象)
    # 這裡的公式應與您 Environment 裡的 r_sec 計算邏輯盡量一致
    h_gain = 1.0 / (d_tar + 1.0)**2
    g_gain = 1.0 / (min_d_eve + 1.0)**2
    # 模擬推理模式下的最優速率 (假設 Beamforming 抵銷了大部分 Eve 干擾)
    curr_r = max(0, 25.0 * h_gain - 5.0 * g_gain + 10.0)
    path_r_secs.append(curr_r)
    last_r_sim = curr_r # 更新回饋給大腦

    # 3. 軌跡物理邏輯
    att_mag = 0.15 * (1 - np.exp(-d_tar/5))
    attraction = (target_pos - uav_p) / (d_tar + 1e-9) * att_mag
    brain_decision = out[0:2] * 0.5
    move_vec = attraction + brain_decision

    # 4. 步長控制
    if d_tar < 5.0 and np.dot(attraction, brain_decision) < 0:
        uav_p += move_vec * 0.2
    else:
        uav_p += move_vec * 0.5

    traj.append(uav_p.copy())
    if d_tar < 0.2: break

# --- 繪圖邏輯 ---
traj = np.array(traj[:-1]) # 對齊長度
path_r_secs = np.array(path_r_secs)

# A. 畫出帶有顏色編碼的軌跡點 (顏色隨速率變化)
sc = ax3.scatter(traj[:, 0], traj[:, 1], c=path_r_secs, cmap='viridis',
                 s=25, zorder=3, edgecolors='none', label='UAV Path (R_sec)')
cbar = plt.colorbar(sc, ax=ax3)
cbar.set_label('Instant Secrecy Rate (bits/s/Hz)', fontsize=10)

# B. 畫出平滑連線
ax3.plot(traj[:, 0], traj[:, 1], 'g-', alpha=0.3)

# C. 標記目標、基站與 Eve
ax3.scatter(bs_pos[0], bs_pos[1], c='black', marker='^', s=200, label='BS (Cloud)')
ax3.scatter(target_pos[0], target_pos[1], c='b', marker='s', s=100, label='Target')
for i, e_pos in enumerate(threat_pos_list):
    ax3.scatter(e_pos[0], e_pos[1], c='r', marker='x', s=100, label="Eves" if i==0 else "")

ax3.set_title(f'(C) Control with {len(threat_pos_list)} Eves (Final R={path_r_secs[-1]:.2f})', fontweight='bold')
ax3.set_xlim(-1, 15); ax3.set_ylim(-1, 15)
ax3.legend(loc='lower right', fontsize='small'); ax3.grid(True, ls=':')

plt.tight_layout()
plt.show()

# 儲存
file_name = f'ROLL_Analysis_{len(threat_pos_list)}Eves_RateVis.png'
plt.savefig(os.path.join(nb_path, file_name), dpi=300)

# 儲存結果
file_name = f'ROLL_Analysis_Num{len(threat_pos_list)}Eves.png'
save_path = os.path.join(nb_path, file_name)
plt.savefig(save_path, dpi=300)
```
